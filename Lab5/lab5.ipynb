```python
import json
import requests

base_url = "https://www.loc.gov/"
LCCNs = [98508155, 2017843202, 2020722343]

itemEndpoint = 'item/'
searchEndpoint = 'search/'
parameters = {'fo' : 'json'}
```

```python
for lccn in LCCNs:
    r = requests.get(
        base_url + itemEndpoint + str(lccn) + "/",
        params=parameters,
    )
    r.raise_for_status()
    print(r.url)

    with open(f"{lccn}.json", "w", encoding="utf-8") as file:
        json.dump(r.json(), file)

    print(f"Saved {lccn}.json")
```

```python
print(r.url)
print ('Status code', r.status_code)
```
