```python
import requests
import json
import csv
```

```python
#Lab5 item ids
item_ids = ['cph.3f05183', 'fsa.8d24709', 'highsm.64003']
```

```python
#setup api call info
base_url = "https://www.loc.gov/"
itemEndpoint = '/resource/'
search = '/search/'

params = {
    'fo': 'json'
}
```

```python
r = requests.get(base_url + itemEndpoint + item_ids[0], params=params)
```

```python
r.status_code
```

```python
r.json().keys()
```

```python
item_metadata = r.json()['item']
```

```python
for k in item_metadata.keys():
    print(k, ":", item_metadata[k])
```

```python
item_ids = ['cph.3a07680', 'cph.3a07681', 'cph.3a07682']
```

```python
for id in item_ids:
    r = requests.get(base_url + itemEndpoint + id, params = params)
    print(r.url)
    print(r.status_code)
    item_data = r.json(['item'])

    with open(id + '_metadata.json' "w", encoding="utf-8") as file:
        f.write(json.dumps(item_data, indent=2))

    print(f"Saved {id}.json")
```
