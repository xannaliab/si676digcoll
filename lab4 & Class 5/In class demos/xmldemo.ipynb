```python
from pathlib import Path
DATA = Path('..','data')
OUT = Path('..', 'output')
EAD_sample = Path("sample-ead-superior-full.xml")

import xml.etree.ElementTree as ET

tree = ET.parse(EAD_sample)
root = tree.getroot()
```

```python
ns = {
    'ead' : 'http://ead3.archivists.org/schema/'
}
ET.register_namespace('', 'http://ead3.archivists.org/schema/')
```

```python
control2 = root.find("ead:control", ns)
print(control2.tag)
```

```python
control = tree.find('ead:control', namespaces=ns)
control.attrib
```

```python
#using xpath in .findall()
#scopecontent
for note in tree.findall('//ead:scopecontent', ns):
    print(note.tag)
```

```python
for note in tree.findall('.//ead:scopecontent', ns):
    print(note.tag)
```

```python
#using xpath in .findall()
#scopecontent
for note in tree.findall('//ead:c', ns):
    print(note.tag)
```

```python
#using xpath in .findall()
#scopecontent
for note in tree.findall('.//ead:c', ns):
    print(note.tag)
```

```python
c_count = 0

for component in tree.findall('.//ead:c', ns):
    c_count += 1
    print(component.tag, ':' , component.attrib)

print(f'Found {c_count} component(s).')
```

```python
#xpath just for series or file
c_count = 0

for component in tree.findall('.//ead:c[@level="series"]', ns):
    c_count += 1
    print(component.tag, ':' , component.attrib)

print(f'Found {c_count} component(s).')
```
